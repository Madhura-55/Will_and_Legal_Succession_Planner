```python
from pydantic import BaseModel
from typing import Optional, List

class Asset(BaseModel):
    asset_type: str          # "immovable" | "movable" | "business"
    description: str
    identifier: Optional[str] = None   # survey no, account no
    estimated_value: Optional[float] = None

class Beneficiary(BaseModel):
    name: str
    relationship: str
    age: int
    allocation: str          # "50%" or "Flat 3B, Adyar"
    is_minor: bool = False

class WillData(BaseModel):
    testator_name: Optional[str] = None
    testator_age: Optional[int] = None
    testator_address: Optional[str] = None
    religion: Optional[str] = None      # determines ISA vs HSA
    family_type: Optional[str] = None   # "nuclear" | "huf" | "single_parent"
    assets: List[Asset] = []
    beneficiaries: List[Beneficiary] = []
    executor_name: Optional[str] = None
    executor_relationship: Optional[str] = None
    witnesses: List[str] = []
    minor_guardian: Optional[str] = None
    residuary_beneficiary: Optional[str] = None
```
